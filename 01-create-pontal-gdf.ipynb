```python
### solution to coordinate system key error lon_0
import cartopy.crs as ccrs

### save cartopy original version once
if not hasattr(ccrs.PlateCarree, '_original_proj4_params'):
    ccrs.PlateCarree._original_proj4_params = ccrs.PlateCarree.proj4_params

def _proj4_params_with_lon_0(self):
    params = ccrs.PlateCarree._original_proj4_params.fget(self)
    params.setdefault('lon_0', params.get('pm', 0))
    return params

ccrs.PlateCarree.proj4_params = property(_proj4_params_with_lon_0)
```

```python
## create geodataframe of pontal region using polygon coordinates (geojson.io)

### set polygon coordinates for pontal region
pontal_coords = [ 
    (-53.1670106, -22.6950238),
    (-51.110509, -22.7213722),
    (-50.9370196, -21.8211815),
    (-52.1795634, -21.4244243),
    (-53.2322963, -22.6272474),
    ( -53.1670106, -22.6950238)
]

### convert polygon to geodataframe
pontal_gdf = gpd.GeoDataFrame(

    ### establish polygon
    geometry = [Polygon(pontal_coords)],

    ### set coordinate reference system
    crs = 'EPSG:4328'
)

### check gdf geomtery for accuracy
pontal_gdf
```

```python
### Create a sitemap using gdf and a basemap of Pontal region
pontal_map = pontal_gdf.hvplot(

    ### set basemap
    geo = True, tiles = 'EsriImagery',

    ### set transparency
    fill_color = 'none',

    ### dispaly polygon regional boundary
    line_color = 'white', line_width = 3,

    ### set figure size
    frame_width = 800, frame_height = 600,

    ### set title
    title = "Pontal do Paranapanema, São Paulo, Brazil"
)

pontal_map
```
